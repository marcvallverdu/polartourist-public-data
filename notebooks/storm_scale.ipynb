# Kp is a label, not a trip guarantee
Five NOAA scale rows; no observations or sunspot values. NOAA facts public domain; source https://www.spaceweather.gov/noaa-scales-explanation ; layout/explanation PolarTourist CC BY 4.0. No endorsement.
Study the 9−/9 distinction. Cloud, darkness and location still matter. See README and RIGHTS.md.

```python
from pathlib import Path
import csv, json
candidates = [Path("data/noaa_geomagnetic_scale.csv"), Path("../data/noaa_geomagnetic_scale.csv")]
if Path("/kaggle/input").exists():
    candidates += list(Path("/kaggle/input").rglob("noaa_geomagnetic_scale.csv"))
path = next((p for p in candidates if p.exists()), None)
assert path is not None, "Attach the supplied NOAA scale CSV"
with path.open(newline="") as f:
    rows = list(csv.DictReader(f))
assert len(rows) == 5
assert [(r["level"], r["kp_display"]) for r in rows] == [("G1","5"),("G2","6"),("G3","7"),("G4","8 or 9−"),("G5","9")]
for r in rows:
    print(f"{r['level']:3} {r['label']:10} Kp {r['kp_display']}")
print("G4 includes 9−; exact 9 is G5. No sighting probabilities inferred.")
```
